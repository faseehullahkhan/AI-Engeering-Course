```python
name1 = "Faseeh"
name2 = "dost"

salary1 = 80000
salary2 = 90000

def calculate_bonus(Salary):
    return Salary * 0.10

print(f"{name1}'s salary is {salary1} and the bonus is {calculate_bonus(salary1)}")
print(f"{name2}'s salary is {salary2} and the bonus is {calculate_bonus(salary2)}")
```

```python
class employee:
    def __init__(self , name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        return self.salary * 0.10

emp1 = employee("Faseeh", 100000)
print(emp1.name)
emp1.calculate_bonus()
```

```python
# Another exmaple
from datetime import date

class human:
    def __init__(self, name , birth_year):
        self.name = name
        self.birth_year = birth_year

    def getAge(self):
        current_year = date.today().year
        return current_year - self.birth_year 
    
    def introduce(self): 
        print(f"hello my name is {self.name} and my age is {self.getAge()} , how are you?")
h1 = human("Faseeh" , 2002)
h1.getAge()
h1.introduce()
```

```python
class YoutubeChannel:
    def __init__(self , name):
        self.name = name
        self.subscriber = 0

    def subscribe(self):
        print("Thanks for subscribing")
        self.subscriber = self.subscriber + 1
        return self.subscriber

    def unsubscribe(self):
        print("So sad to see you gooooooo!!!")
        self.subscriber = self.subscriber - 1
        return self.subscriber

    def show(self):
        print(f"Channel name is {self.name} and count of the subscribers are {self.subscriber}")

y1 = YoutubeChannel("Autumn Wells")
y1.subscribe()
y1.subscribe()
y1.subscribe()
```
