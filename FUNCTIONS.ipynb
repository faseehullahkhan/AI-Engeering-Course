```python
def calculate_profit_margin(revenue, expenses):
    profit = revenue - expenses
    margin = (profit * 100) / revenue
    return profit , margin
revenue = 88 
expenses = 74
profit , margin = calculate_profit_margin(revenue , expenses)
print(f"profit is {profit} and margin is {margin:.2f}") 
```

```python
# It will return the answer in the tuple format
def names_of_people(*args):
    print(*args)

names_of_people(10 , 20 , 40)
```

```python
# But there is another good way then *args, which is
def people_names(**kargs):
    print(kargs)
    # or we can run a for loop
    for key , value in kargs.items():
        print(value)

people_names(name1 = "Faseeh" , name2 = "Ahmed" , name3 = "Saif")
```
