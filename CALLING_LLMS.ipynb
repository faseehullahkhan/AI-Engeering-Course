```python
def generate_recipe(ingradients , cuisine , diet):
    '''
        Generate a recipe using ingradients provided.
        cuisine can be "italian" , "maxican" , "Pakistani" , "Any" etc.
        diet can be "Vegan" , "Vegetarian" , "non-vegetarians" , "Any" etc.
    '''
    prompt = f"Generate one food recipe using these ingradients: {',' .join(ingradients)}Recipe should not be more then 100 wordscuisine: {cuisine}diet: {diet}"
    response = client.models.generate_content(
    model="gemini-3.5-flash", 
    contents=prompt ) 
    return response.text
```

```python
recipe  = generate_recipe(["Potato" , "Meat" , "salt" , "green coriander"] , "Pakistani" , "non-vegan")
print(recipe)
```
