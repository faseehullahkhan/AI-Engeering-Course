LISTS


```python
revenues = [66 , 89 , 90 ]
expenses = [40 , 50 , 55]
print(revenues[0:len(revenues):2])
```

```python
# Now to calculate the profit margin for each quarter lets say we can do this
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = (profit * 100) / expenses[i]
    print(profit)
    print(margin)
```

DICT

```python
profits_revenues = {
    "revenue" : 80,
    "expenses" : 20
} 

profits_revenues["profit"] = profits_revenues["revenue"] - profits_revenues[ "expenses" ]
print(profits_revenues )
```

```python
charges = {
     "Q1" : {"revenue" : 80, "expenses" : 20},
     "Q2" : {"revenue" : -20, "expenses" : 20},
     "Q3" : {"revenue" : 60, "expenses" : 20}
}


for key , value in charges.items():
    if value["revenue"] < 0:
        print(f"Invalid revenue for {key}... SKIPPING")
        continue
    print(f"quarter : {key} = profit = {value}")
```

```python
students = {
    "Faseeh" : [85, 90 , 78],
    "sami" : [72, 68 , 75],
    "ahmed" : [45, 52 , 48],
    "hasnain" : [95, 92, 98],
    "obaid" : [60 , 65 ,70]    
}
```

```python
marks = []
for key , value in students.items():
    avg = sum(value)//3
    marks.append(avg)
    if(avg >= 80):
        print(f"TOPPER OF THE CLASS is {key} and the average is {avg}")
        
    elif(avg > 60 and avg <= 79):
        print("PASS")
    else:
        print("NEEDS IMPROVEMENT")
class_avg = sum(marks)//len(marks)
print(f"the class average score is = {class_avg}")
```
